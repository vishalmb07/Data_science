```python
company =['TCS','IBM','EY','Techm','INFY']

share=[250,300,100,80,400]
```

```python
plt.bar(company,share,label=2026,width=0.4)
plt.xlabel('Comapny')
plt.ylabel('share')
plt.legend()
```

```python
#subplot

plt.subplot(2,2,1)
plt.bar(company,share,label=2026,width=0.4)
plt.xlabel('Comapny')
plt.ylabel('share')
plt.xticks(rotation=20)
plt.legend()

plt.subplot(2,2,2)
plt.pie(share,labels=company,autopct='%.f%%',colors=['y','c','r','b','m'])
plt.show()

plt.subplot(2,2,3)
plt.hist(salary,bins=[10000,25000,30000,50000,60000,75000,90000],label=2026,color='g')
plt.legend()


plt.subplot(2,2,4)
plt.scatter(range(100),data,marker='+',color='m')
```

```python
#subplot

plt.subplot(2,2,1)
plt.bar(company,share,label=2026,width=0.4)
plt.xlabel('Comapny')
plt.ylabel('share')
plt.xticks(company,['tcs','ibm','ey','tm','inf'],rotation=20)
plt.legend()

plt.subplot(2,2,2)
plt.pie(share,labels=company,autopct='%.f%%',colors=['y','c','r','b','m'])
plt.show()

plt.subplot(2,2,3)
plt.hist(salary,bins=[10000,25000,30000,50000,60000,75000,90000],label=2026,color='g')
plt.legend()


plt.subplot(2,2,4)
plt.scatter(range(100),data,marker='+',color='m')
```
