```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'State': ['Maharashtra', 'Gujrat', 'Utter pradesh', 'Himachal Pradesh'],
    'Count': [140, 90, 100, 400]
})



df.plot(
    x='State',
    y='Count',
    kind='bar',
    width=0.4,
    color='#438C1D'
)

plt.title('State Vs Count')
plt.xlabel('States of India')
plt.ylabel('Count')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

state = ['Maharashtra', 'Gujrat', 'Utter pradesh', 'Himachal Pradesh']

record = [140, 90, 100, 400]
record_2026 = [280, 250, 100, 200]

x = np.arange(len(state))
width = 0.35

plt.figure(figsize=(6, 4))

# 2025 bar
plt.bar(x - width/2, record,
        width=width,
        color='#438C1D',
        label='2025')

# 2026 bar
plt.bar(x + width/2, record_2026,
        width=width,
        color='blue',
        label='2026')

plt.title('State Vs Count')

plt.xlabel('States of India')
plt.ylabel('Count')

plt.xticks(x, state, rotation=45)

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(3,3))
plt.plot(state,record,color='#438C1D',label='2025')
plt.legend()
plt.show()

plt.figure(figsize=(3,3))
plt.plot(state,[280,250,100,200],color='red',label='2026')
plt.legend()
plt.show()
```
