# **Data Visualization using Seaborn**

https://seaborn.pydata.org/index.html?pubDate=20251004

Seaborn is a Python data visualization library based on matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics.

```python
import seaborn as sns
```

```python
# seaborn has built in dataset
# load datasets
print(sns.get_dataset_names())
```

```python
# lets use dataset
p = sns.load_dataset('penguins')
p.head(2)
```

```python
p.shape
```

```python
# count per species
import matplotlib.pyplot as plt
plt.figure(figsize=(3,3))
sns.countplot(data=p,x='island',hue='sex',palette='rainbow')
plt.show()
```

```python
p[:1]
```

```python
sns.histplot(data=p,x='body_mass_g',bins=[2500,3000,3500,4000],hue='sex')
```

### Conclusion:
1. In the range 2000-3000 we dont have Male penguin. All are femake
2. Male penguin weight is more, goes above 3000
3. In 3000 to 3500 We have more Female penguins than Male

```python
# Category scatter plot
sns.stripplot(data=p,x='species',y='body_mass_g',hue='sex')
```

```python
p[:1]
```

```python
# scatter plot
sns.scatterplot(data=p,x='bill_length_mm',y='bill_depth_mm',hue='sex')
```

```python
# box plot gives u statistical knowledge
sns.boxplot(data=p['bill_depth_mm'],orient='h')
```
